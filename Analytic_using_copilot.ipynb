```python
import pandas as pd

# 1. Load CSV
df = pd.read_csv("/content/drive/MyDrive/MTIT/Employe_Performance_dataset.csv")

# 2. Remove duplicates
df = df.drop_duplicates()

# 3. Handle missing values
# Option A: Drop rows with missing values
df = df.dropna()

# Option B: Fill missing values (example: fill numeric with mean, categorical with mode)
# Replace 'Age' and 'Gender' with actual column names from your dataset if different.
# If Option A (dropna) is executed, there might be no missing values left to fill.

# Fill missing values for a numeric column (e.g., 'Age')
if 'Age' in df.columns:
    df['Age'] = df['Age'].fillna(df['Age'].mean())

# Fill missing values for a categorical column (e.g., 'Gender')
if 'Gender' in df.columns:
    df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

# 4. Convert datatypes
# If you have a date column, uncomment the line below and replace 'your_date_column' with the actual column name
# df['your_date_column'] = pd.to_datetime(df['your_date_column'], errors='coerce')

# Ensure a numeric column (e.g., 'Age') is of numeric type
if 'Age' in df.columns:
    df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# 5. Show summary statistics
print(df.describe(include='all'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# The DataFrame 'df' is already loaded and processed in previous cells.
# Remove the line below as it tries to load a non-existent file and is redundant.
# df = pd.read_csv("your_file.csv")

# Histogram for a numeric column (e.g., 'Age')
plt.figure(figsize=(8,6))
sns.histplot(df['Age'], bins=30, kde=True)
plt.title("Histogram of Employee Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Bar Chart for a categorical column (e.g., 'Gender')
plt.figure(figsize=(8,6))
sns.countplot(x='Gender', data=df)
plt.title("Bar Chart of Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()

# Correlation Heatmap for relevant numeric columns
plt.figure(figsize=(10,8))
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=['number'])
corr = numeric_df.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()

# Boxplot of a numeric column by a categorical column (e.g., 'Salary' by 'Department')
plt.figure(figsize=(8,6))
sns.boxplot(x='Department', y='Salary', data=df)
plt.title("Boxplot of Salary by Department")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()
```
