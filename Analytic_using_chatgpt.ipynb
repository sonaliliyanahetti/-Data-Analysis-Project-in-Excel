```python
import pandas as pd

# Load CSV
df = pd.read_csv('/content/drive/MyDrive/MTIT/Employe_Performance_dataset.csv')  # replace with your CSV path

# Remove duplicates
df = df.drop_duplicates()

# Handle missing values
# Option A: Drop rows with any missing values
df = df.dropna()

# Option B: Fill missing values (example: numeric columns with mean, categorical with mode)
for col in df.select_dtypes(include=['float', 'int']):
   df[col] = df[col].fillna(df[col].mean())
for col in df.select_dtypes(include=['object']):
    df[col] = df[col].fillna(df[col].mode()[0])

# Convert datatypes (example: convert 'date_column' to datetime, 'id' to int)
# The 'date_column' does not exist in the DataFrame, so this line is removed/commented out.
df['ID'] = df['ID'].astype(int) # Corrected 'id' to 'ID'

# 5 Show summary statistics
print(df.info())        # info about columns and datatypes
print(df.describe())    # numeric summary statistics
print(df.describe(include='object'))  # categorical summary
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# The DataFrame 'df' is already loaded and preprocessed from previous cells.
# Removing the line that attempts to load 'your_file.csv'.

# ------------------------
# 1️⃣ Histogram (for numeric columns)
# ------------------------
numeric_cols = df.select_dtypes(include=['int', 'float']).columns
for col in numeric_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(df[col], bins=30, kde=True, color='skyblue')
    plt.title(f'Histogram of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# ------------------------
# 2️⃣ Bar Chart (for categorical columns)
# ------------------------
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    plt.figure(figsize=(8,4))
    # Fix: Explicitly assign hue to the same column as x for future compatibility
    sns.countplot(data=df, x=col, hue=col, palette='viridis', order=df[col].value_counts().index, legend=False)
    plt.title(f'Bar Chart of {col}')
    plt.xlabel(col)
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.show()

# ------------------------
# 3️⃣ Correlation Heatmap (numeric columns only)
# ------------------------
plt.figure(figsize=(10,8))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()

# ------------------------
# 4️⃣ Boxplot (numeric vs categorical)
# Example: plot numeric column distributions grouped by a categorical column
# ------------------------
if categorical_cols.any() and numeric_cols.any():
    for cat_col in categorical_cols:
        for num_col in numeric_cols:
            plt.figure(figsize=(8,4))
            # Fix: Explicitly assign hue to the same column as x for future compatibility
            sns.boxplot(x=cat_col, y=num_col, data=df, hue=cat_col, palette='Set2', legend=False)
            plt.title(f'Boxplot of {num_col} by {cat_col}')
            plt.xticks(rotation=45)
            plt.show()
```
